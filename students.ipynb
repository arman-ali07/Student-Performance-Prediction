```python
plt.figure(figsize=(8,5))
sns.histplot(df["GPA"], kde=True, bins=20)

plt.title("GPA Distribution")
plt.xlabel("GPA")
plt.ylabel("Number of Students")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 6))
sns.histplot(df["StudyTimeWeekly"], kde=True, bins=20)

plt.title("Study Distribution")
plt.xlabel("Study hours per week")
plt.ylabel("Number of Stundets")
plt.xlim(0,20)
plt.margins(x=0.20)

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8,6))
sns.countplot(x="Gender", data=df)

plt.title("Gender Distibution")
plt.xlabel("Gender")
plt.ylabel("Number of Counts")
plt.xticks([0,1], ["Male","Female"])
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8,6))
sns.countplot(x="GradeClass", data=df)

plt.title("Class Grade Distribution")
plt.xlabel("Class Grade")
plt.ylabel("Number of Studnets")
plt.tight_layout()
plt.show()
```
